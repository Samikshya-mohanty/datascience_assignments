Q1. For the suburb of Altona, it is postulated that a typical property sells for $800,000.


Use the data at hand to test this assumption.
Is the typical property price really $800,000 or has it increased?
Use a significance level of 5%.

Null hypothesis: mean <= $800,000

Alternate Hypothesis: mean > $800,000

For this we will do one sample right tailed T - test.

```python
df_altona = df[df['Suburb'] == 'Altona']
```

```python
df_altona
```

We have filtered the data for the suburb of Altona and got 74 samples.

```python
df_altona['Price'].describe()
```

T test formula

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$$

```python
sample_mean = df_altona['Price'].mean()
sample_std_deviation = df_altona['Price'].std()
sample_size = len(df_altona)
null_hypothesis = 800000
```

```python
t_value = (sample_mean - null_hypothesis) / (sample_std_deviation / mth.sqrt(sample_size))
t_value
```

```python
t_stat, p_value = stats.ttest_1samp(df_altona['Price'], null_hypothesis, alternative='greater')
```

```python
t_stat
```

```python
p_value
```

As the P-value 0.15 is greater than significant value 0.05, hence we fail to  reject the null hypothesis.

There is not enough evidence to claim that the price has increased beyond $\$800,000$.

Q2. For the year 2016, is there any difference in the prices of properties sold in the
summer months vs winter months?

• Consider months from October till March as winter months and rest as
summer months.

• Use a significance level of 5%.


Null Hypothesis : Avg price of summer_mean = Avg price of winter_mean

Alternate Hypothesis :  Avg price of summer_mean != Avg price of winter_mean

Independent Two-Sample T-test

```python
df.head()
```

```python
# Filtering the data for the year 2016
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
df_2016 = df[df['Date'].dt.year == 2016]
```

```python
df_2016
```

```python
# separating winter and and summer months prices
winter_months = [10, 11, 12, 1, 2, 3]
winter_prices = df_2016[df_2016['Date'].dt.month.isin(winter_months)]['Price'].dropna()
summer_prices = df_2016[~df_2016['Date'].dt.month.isin(winter_months)]['Price'].dropna()
```

```python
def get_metrics(df:pd.Series):
  sample_mean = df.mean()
  sample_std_deviation = df.std()
  sample_size = len(df)
  return sample_mean, sample_std_deviation, sample_size
```

```python
summer_mean, summer_std_deviation, summer_sample_size = get_metrics(summer_prices)
summer_mean, summer_std_deviation, summer_sample_size
```

```python
winter_mean, winter_std_deviation, winter_sample_size = get_metrics(winter_prices)
winter_mean, winter_std_deviation, winter_sample_size
```

$$t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

```python
t_value = (summer_mean - winter_mean) / mth.sqrt((summer_std_deviation**2 / summer_sample_size) + (winter_std_deviation**2 / winter_sample_size))
t_value
```

```python
t_stat, p_value = stats.ttest_ind(winter_prices, summer_prices, equal_var=False, alternative='two-sided')
```

```python
t_stat, p_value
```

Since the p-value ($0.0001$) is significantly lower than the threshold of $0.05$, we reject the null hypothesis.

And, we conclude that, there is a significant difference in the prices of properties sold in the summer months vs the winter months in 2016

Q7. Independent Two-Sample T-Test (Feature Impact)
Do properties with car parking sell at a higher average price than properties without
car parking, across the entire dataset? Use a 5% significance level and justify:

• Choice of test

• Interpretation of p-value

• Business implications for developers

Null hypthesis: avg_price_car_parking > avg_price_nocar_parking

Alternate hypthesis: avg_price_car_parking < avg_price_nocar_parking

Significance = 0.05

```python
# with car parking
df_car_parking = df[df['Car']>0]

# without car parking
df_nocar_parking = df[df['Car']==0]
```

```python
with_car_mean,with_car_std,with_car_sample_size = get_metrics(df_car_parking['Price'])
without_car_mean,withoout_car_std,without_car_sample_size = get_metrics(df_nocar_parking['Price'])
```

```python
t_value = (with_car_mean - without_car_mean) / mth.sqrt((with_car_std**2 / with_car_sample_size) + (withoout_car_std**2 / without_car_sample_size))
t_value
```

```python
t_stat, p_value = stats.ttest_ind(df_car_parking['Price'], df_nocar_parking['Price'], equal_var=False, alternative='greater')
```

```python
t_stat, p_value
```

As p-value 0.607 > 0.05, we fail to reject the Null Hypothesis.

We can conclude Developers should prioritize parking as it adds measurable value.

Q8. Two-Way ANOVA (Location & Property Type)
Investigate whether property prices are influenced by:

- Suburb

- Type of property

- Interaction between suburb and property type

Use Two-Way ANOVA and explain which factors significantly affect price

null hypothesis

H0: There is no influence on the property price between different property types.

H1 : There is no influence on the property price between different Suburb.

H2 : There is no interaction between suburb and property type.

Alternate hypothesis:

HA0: There is influence on the property price between different property types.

HA1 : There is influence on the property price between different Suburb.

HA2 : There is some interaction between suburb and property type.






```python
df_annnova = df.groupby(by=['Type','Suburb'])['Price'].agg(['count','mean'])
df_annnova
```

```python
df_cleaned_suburb = df['Suburb'].dropna()
df_cleaned_type = df['Type'].dropna()
df_cleaned_price = df['Price'].dropna()
```

```python
df_cleaned = pd.DataFrame({'Suburb':df_cleaned_suburb, 'Type':df_cleaned_type, 'Price':df_cleaned_price})
```

```python
df_cleaned.head()
```

```python
model = ols('Price ~ C(Type) + C(Suburb) + C(Type):C(Suburb)', data=df_cleaned).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

As the p-value is significantly less than 0.05 for all the three hypothesis, hence rejecting the null hypothesis.

So we can conclude that type of property and suburb affect the price of the property and there is an interaction between suburb and type of property.
